## Transform Hero Features

```python
df_hero_features = pd.read_sql(
    "SELECT * FROM hero_features",
    con=engine
)

df_hero_features
```

### One-Hot-Encoding

```python
from sklearn.preprocessing import MultiLabelBinarizer
```

```python
from constants.constants import HERO_DICT
```

```python
ALL_HEROES = list(HERO_DICT.values())
ALL_HEROES
```

```python
mlb = MultiLabelBinarizer(classes=ALL_HEROES)
hero_matrix = mlb.fit_transform(df_hero_features['hero_picks'])
hero_features = pd.DataFrame(hero_matrix, columns=mlb.classes_)
hero_features.insert(0, 'match_id', df_hero_features['match_id'].values)
```

### Player Hero Features

```python
df_player_hero_features = pd.read_sql(
    "SELECT * FROM player_hero_features",
    con=engine
)

df_player_hero_features
```

## Combined Features

```python
merged_df = pd.merge(
    hero_features,
    df_player_hero_features,
    on='match_id',
    how='inner'
)

final_df = pd.merge(
    merged_df,
    df_team_features,
    on='match_id',
    how='inner'
)

final_df
```
