```python
import asyncio
import time
```

```python
# Async functions objects are blue-print for creating coroutine objects
async def add_numbers(a, b):
    await asyncio.sleep(3)
    return a + b

# Creates a coroutine object when async function object is called, similar to instantiation in OOP. 
# Coroutine object which returns the actual results, is not called. 
coroutine =  add_numbers(1, 3)

# Call the funciton using await to return the result
await coroutine 
```

```python
task_dict = {
    'number_1': asyncio.create_task(add_numbers(1, 3)),
    'number_2': asyncio.create_task(add_numbers(2, 4))
}

task =  await asyncio.gather(*task_dict.values())

```
