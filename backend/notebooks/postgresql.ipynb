```python
import psycopg2
from psycopg2 import Error
from sqlalchemy.engine import create_engine, URL
from sqlalchemy import text
import pandas as pd

url_object = URL.create(
    "postgresql+psycopg2",
    username='liuhaochen',
    host='localhost',
    port='5432',
    database='test',
    password='110799'
)

engine = create_engine(url_object)


try:
    # Establish a connection using the engine
    with engine.connect() as connection:
        # Execute the query using the connection object
        result = connection.execute(text("SELECT * FROM pro_matches"))

        # Fetch all the results and load them into a DataFrame
        sql_df = pd.DataFrame(result.fetchall(), columns=result.keys())

except Exception as e:
    print(f"Error fetching data from the database: {e}")
```
