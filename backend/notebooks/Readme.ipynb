# Dota 2 Professional Match Predictor 

## Aim of project
1. To use machine learning to predict the outcome of a professional Dota 2 match.
2. Determine how much bets is to be placed on the winning team



# 1. Types of Dota 2 Betting

### 1. Pre-match Betting 
To bet on the outcome of a match before the event begins. Information available at this stage usually comprise of:
   1. Teams involved
   2. Players involved

### 2. Live Betting 
Live Betting is a lot more complicated. Bets can be made on various outcome, including but not limited to 
1. The outcome of the match
2. First blood (The first kill of the game)
3. First Roshan Kill
4. First to 10 kills 

A lot more information is available at this stage, consisting of:
1. Teams involved
2. players involved
3. Heros selected by each player on each team
4. Tower and barrack status
5. Team score, experience and gold differences
6. Hero's build, item, gold and levels e.t.c. 

# 2. Data Gathering

## 2.1 OpenDota API

### Pros
1. Data fetched from the OpenDota's API tends to be cleaner and more properly structured

### Cons:
1. Limited to 50000 api calls a month
2. 60 calls per limit
3. Public and Pro matches only return the most recent 100 matches - unable to fetch large number of training data 

## 2.2 Steam's Web API
### Pros:
1. 100,000 calls per day

### Cons:
1. get_match_history endpoint only return the most recent 500 matches - unable to fetch large number of training data (>10000)

### Solution to collect large amount of training data
1. Use pro_players endpoint, which returns a list of all the currently registered pro players to retrieve their account_id
2. For each account_id retrieve a list of match_ids of the current patch 
3. Then get the match details of each match using the Steam Web API
